##Read each CSV file and append to the list which will store DataFrames
##Concatenate all DataFrames into a single DataFrame

```python
import pandas as pd
import os

# The path to the dataset directory
dataset_directory = '/kaggle/input/12-months-sales-data'

# List files in the directory
files_in_directory = os.listdir(dataset_directory)
print("Files in the dataset directory:", files_in_directory)

# List to store DataFrames
all_months_data = []

# Read each CSV file and append to the list
for file in files_in_directory:
    if file.endswith('.csv'):
        csv_file_path = os.path.join(dataset_directory, file)
        try:
            df_month = pd.read_csv(csv_file_path)
            all_months_data.append(df_month)
        except Exception as e:
            print(f"Error reading {file}: {e}")

# Concatenate all DataFrames into a single DataFrame
if all_months_data:
    df = pd.concat(all_months_data, ignore_index=True)
    print("\nAll datasets loaded and combined successfully:")
    display(df.head())  # Display the first few rows
    display(df.tail())  # Display the last few rows
else:
    print("\nNo CSV files found or could not be read in the dataset directory.")
```
