###Checking weather data is loaded using head function

```python
import pandas as pd

df=pd.read_csv('/root/.cache/kagglehub/datasets/prepinstaprime/black-friday-sales-data/versions/1/train.csv')
df.head()
```

###Filtering the DataFrame to find all purchases with a value higher than $15,000

```python
high_value_purchases = df[df['Purchase'] > 15000]
print(high_value_purchases)
```

###Combine multiple conditions to find a specific customer segment
###Then apply all conditions simultaneously using the '**&**' operator

```python
condition1 = df['Gender'] == 'F'
condition2 = df['City_Category'] == 'A'
condition3 = df['Age'] == '0-17'

target_segment = df[condition1 & condition2 & condition3]
display(target_segment)
```
