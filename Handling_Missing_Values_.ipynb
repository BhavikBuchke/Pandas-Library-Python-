### **1. Import required modules (files from google.colab and pandas)**
### **2. Upload the file**
### **3. printing the uploaded file's name**

```python
from google.colab import files
import pandas as pd

uploaded = files.upload()
Titanic = list(uploaded.keys())[0]

print(f"File '{Titanic}' uploaded successfully.")
```

### **Load the data from the file into a pandas DataFrame using**
### **exception hadling to avoide error**



```python
try:
    df = pd.read_csv(Titanic)
    print("Data loaded into a pandas DataFrame:")
    display(df.head())
except Exception as e:
    print(f"Error loading data: {e}")
```

### **Inspecting dataset for Empty/Null values**

```python
df.isnull()
```

### **Empty values based on Columns**

```python
df.isnull().sum()
```

### **Total no. of empty values in dataset**

```python
df.isnull().sum().sum()
```

### **Handling missing values in the dataset**
1. Removing the column that contains too many missing values
2. Filling Age & Fare column with mean of values/entries
3. Print the result after handling the empty/null values



```python
#df = df.drop('Cabin', axis=1)
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Fare'] = df['Fare'].fillna(df['Fare'].mean())
print("Missing values after handling:\n",df.isnull().sum())
```
