###Importing Data from Kaggle


```python
import kagglehub
import os

# Download a specific file from the latest version
path = kagglehub.dataset_download("datafiniti/womens-shoes-prices", "Datafiniti_Womens_Shoes.csv")

print("Path to the file:", path)

# Assuming the file is a CSV, read it into a pandas DataFrame
import pandas as pd

if os.path.exists(path):
    df = pd.read_csv(path)
else:
    print(f"Error: {path} not found.")
```

###Sort the DataFrame by the price column in descending order to place the most expensive items at the top

```python
sorted_df = df.sort_values(by='prices.amountMax', ascending=False)
```
