###Load the data

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("skmewati/myntra-sales-dataset")

print("Path to dataset files:", path)
```

```python
import pandas as pd
df=pd.read_csv(path+"/myntra_dataset_ByScraping.csv")
df.head()
```

###Select a text column and applying lowercasing to standardize the text

```python
df['pants_description'] = df['pants_description'].str.lower()
df.head()
```

###Remove a common but uninformative word, such as **"men's,"** from the text to clean it further

```python
df['pants_description'] = df['pants_description'].str.replace("men's", "")
df.head()
```
