###Load data

```python
import kagglehub
path = kagglehub.dataset_download("ibrohimsharipov/12-months-sales-data")

print("Path to dataset files:", path)
```

```python
import pandas as pd
df=pd.read_csv(path+"/Sales_April_2019.csv")
df.head()
```

###Convert the **'Order Date'** column from a string to a datetime object

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%m/%d/%y %H:%M', errors='coerce')
df.head()
```

###Create a new **'Month'** column by extracting the month number from the **'Order Date'** column

```python
df['Month'] = df['Order Date'].dt.month
df.head()
```

###create a **'Year'** column from the 'Order Date' column

```python
df['Year'] = df['Order Date'].dt.year
df.head()
```

###Creating a 'DayOfWeek' column

```python
df['DayOfWeek'] = df['Order Date'].dt.dayofweek
df.head()
```
