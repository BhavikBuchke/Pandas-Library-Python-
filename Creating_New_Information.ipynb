###Load the data

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("ibrohimsharipov/12-months-sales-data")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import os

# Get the path to the downloaded dataset files
path = '/root/.cache/kagglehub/datasets/ibrohimsharipov/12-months-sales-data/versions/1'

# List all files in the directory
all_files = [os.path.join(path, f) for f in os.listdir(path) if f.endswith('.csv')]

# Read and concatenate all CSV files into a single DataFrame
df_list = []
for file in all_files:
    df_list.append(pd.read_csv(file))

df = pd.concat(df_list, ignore_index=True)

# Display the first few rows of the combined DataFrame
df.head()
```

###Drop rows with missing values

```python
# Drop rows where 'Order Date' is 'Order Date' (which indicates a header row)
df = df[df['Order Date'] != 'Order Date']

# Drop rows with missing values in the relevant columns
df.dropna(subset=['Quantity Ordered', 'Price Each'], inplace=True)
df.columns
```

###converte columns like **'Quantity Ordered'** and **'Price Each'** from String to numeric datatype

```python
df['Quantity Ordered'] = pd.to_numeric(df['Quantity Ordered'])
df['Price Each'] = pd.to_numeric(df['Price Each'])

df[['Quantity Ordered', 'Price Each']].dtypes
```

####Creating a new **'Total Sale'** column by multiplying the ***'Quantity Ordered'*** column by the ***'Price Each'*** column

```python
df['Total Sales'] = df['Quantity Ordered'] * df['Price Each']
display(df.head())
```
